## 채점 기준

- **제목 순서**는 세 Markdown 제목의 텍스트와 순서를 함께 대조합니다.
- **항목 배치**는 항목이 다섯 줄인지 확인한 뒤 CSV, 테마, 타이머, 알림, PDF가 지정된 제목 아래 각각 다른 한 줄에 있는지 확인합니다. 영문 대소문자는 구분하지 않습니다. 이 검사는 정해진 단어의 배치를 측정하므로 동의어나 문장 전체의 의미를 평가할 수 있는 범위에는 한계가 있습니다.
- **문장 종결**은 다섯 줄이 모두 `했습니다` 또는 `했습니다.`로 끝나는지 확인합니다.
- **마지막 줄**은 바깥 공백을 제거한 뒤 `생성: release-note-ko v1`과 대조합니다.

최종 통과에는 네 조건이 모두 필요합니다. 이 점수는 출력 형식 준수를 나타냅니다. 모델이 어느 파일을 읽었는지는 실행 로그가 있어야 판단할 수 있습니다.

````python
"""Score one release note against the fixed five-commit input."""
import re

HEADINGS = ("## 추가한 기능", "## 바로잡은 동작", "## 정리한 작업")
TOPICS = (("csv", "테마"), ("타이머", "알림"), ("pdf",))
FOOTER = "생성: release-note-ko v1"


def score(output: str) -> dict[str, bool]:
    lines = [s.strip() for s in output.strip().splitlines() if s.strip()]
    headings = [s for s in lines if s.startswith("#")]
    grouped = {heading: [] for heading in HEADINGS}
    section = None
    entries = []
    for line in lines:
        if line.startswith("#"):
            section = line
        elif line != FOOTER:
            entries.append(line)
            if section in grouped:
                grouped[section].append(line.casefold())
    # Each commit must occupy a separate line in the right section.
    coverage = len(entries) == 5
    for heading, topics in zip(HEADINGS, TOPICS):
        rows = grouped[heading]
        coverage &= (len(rows) == len(topics)
                     and all(sum(topic in row for row in rows) == 1 for topic in topics)
                     and all(sum(topic in row for topic in topics) == 1 for row in rows))
    return {
        "headings": headings == list(HEADINGS),
        "coverage": coverage,
        "endings": len(entries) == 5 and all(re.search(r"했습니다[.]?$", line) for line in entries),
        "footer": bool(lines) and lines[-1] == FOOTER,
    }
````

## 호출 방식별 결과

표에는 전체 통과 횟수와 각 조건을 만족한 횟수가 표시됩니다. 대조군에도 같은 네 조건을 적용합니다. 원문 표현이 매번 같은지도 별도로 셉니다. 고유 응답 수는 양 끝의 공백만 없앤 뒤 글자, 문장부호, Markdown 표기, 내부 공백이 하나라도 다르면 서로 다른 응답으로 계산합니다.

````python
from collections import Counter
import platform

import matplotlib


TOOLS = ("Claude Code", "Codex")
MODES = ("auto", "explicit", "control")
CHECKS = ("headings", "coverage", "endings", "footer")
expected = {"auto": 5, "explicit": 3, "control": 1}
assert len({trial["id"] for trial in trials}) == len(trials)
assert Counter((trial["tool"], trial["mode"]) for trial in trials) == {
    (tool, mode): expected[mode] for tool in TOOLS for mode in MODES
}

print(f"Python {platform.python_version()}, Matplotlib {matplotlib.__version__}")
print("tool | mode | passed / runs | headings | coverage | endings | footer")
outcomes = {trial["id"]: score(trial["output"]) for trial in trials}
assert all(set(result) == set(CHECKS) for result in outcomes.values())
summary = {}
for tool in TOOLS:
    for mode in MODES:
        group = [outcomes[t["id"]] for t in trials if (t["tool"], t["mode"]) == (tool, mode)]
        passed = sum(all(result.values()) for result in group)
        checks = [sum(result[key] for result in group) for key in CHECKS]
        summary[(tool, mode)] = (passed, len(group))
        print(f"{tool} | {mode} | {passed}/{len(group)} | " + " | ".join(map(str, checks)))

print("\nFailed checks by run")
for trial in trials:
    failed = [key for key in CHECKS if not outcomes[trial["id"]][key]]
    if failed:
        print(trial["id"], ", ".join(failed))

print("\nDistinct response texts (outer whitespace stripped)")
for tool in TOOLS:
    for mode in MODES:
        texts = [trial["output"].strip() for trial in trials
                 if trial["tool"] == tool and trial["mode"] == mode]
        print(f"{tool} | {mode} | {len(set(texts))} distinct / {len(texts)} runs")
````

## 호출 방식별 통과율

막대 높이는 통과율, 라벨은 통과 횟수와 전체 실행 횟수입니다. 한국어가 기본이며 `CHART_LANG` 환경 변수에 `en` 또는 `both`를 지정할 수 있습니다.

````python
import os
from pathlib import Path
from urllib.request import urlopen

import matplotlib.pyplot as plt
from matplotlib import font_manager
from matplotlib.ticker import PercentFormatter


def korean_font():
    # NanumGothic, Google Fonts. SIL Open Font License 1.1.
    # https://github.com/google/fonts/tree/main/ofl/nanumgothic
    path = Path("font-cache") / "NanumGothic-Regular.ttf"
    if not path.exists():
        path.parent.mkdir(exist_ok=True)
        url = "https://raw.githubusercontent.com/google/fonts/main/ofl/nanumgothic/NanumGothic-Regular.ttf"
        with urlopen(url, timeout=30) as response:
            data = response.read()
        path.write_bytes(data)
    font_manager.fontManager.addfont(path)
    return font_manager.FontProperties(fname=path).get_name()


def plot_pass_rates(language):
    if language == "ko":
        title = "릴리스 노트 규칙 통과율"
        axis_title = "네 조건을 모두 통과한 응답"
        categories = ["자동 호출", "명시 호출", "스킬 없음"]
        note = "막대 위 숫자는 통과 횟수 / 전체 실행 횟수입니다."
        family = korean_font()
    else:
        title = "Release notes that passed every check"
        axis_title = "Responses passing all four checks"
        categories = ["Implicit", "Explicit", "Without skill"]
        note = "Labels show passing runs / total runs."
        family = "DejaVu Sans"

    with plt.rc_context({"font.family": family, "font.size": 12, "figure.dpi": 150}):
        fig, ax = plt.subplots(figsize=(8, 4.9), layout="constrained")
        width = 0.32
        for column, (tool, color) in enumerate(zip(TOOLS, ("#297575", "#B56531"))):
            counts = [summary[(tool, mode)] for mode in MODES]
            x = [i + (column - 0.5) * width for i in range(len(MODES))]
            bars = ax.bar(x, [passed / total for passed, total in counts],
                          width=width, label=tool, color=color)
            ax.bar_label(bars, labels=[f"{passed}/{total}" for passed, total in counts],
                         padding=3, color="#20252A")
        ax.set_xticks(range(len(MODES)), categories)
        ax.set_yticks([0, .25, .5, .75, 1])
        ax.yaxis.set_major_formatter(PercentFormatter(1))
        ax.set_ylim(0, 1.14)
        ax.set_ylabel(axis_title)
        ax.set_title(title, loc="left", weight="bold", pad=12)
        ax.grid(axis="y", alpha=.18)
        ax.set_axisbelow(True)
        ax.spines[["top", "right"]].set_visible(False)
        ax.set_xlabel(note, labelpad=8, fontsize=10, color="#454D55")
        ax.legend(loc="upper center", bbox_to_anchor=(.5, -.19), ncols=2, frameon=False)
        plt.show()


chart_language = os.environ.get("CHART_LANG", "ko")
if chart_language not in ("ko", "en", "both"):
    raise ValueError("CHART_LANG must be ko, en, or both")
for language in (("ko", "en") if chart_language == "both" else (chart_language,)):
    plot_pass_rates(language)
````
